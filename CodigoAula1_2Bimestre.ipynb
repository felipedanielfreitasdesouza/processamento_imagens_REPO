```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
plt.axis("off")
plt.imshow(data.moon())
```

```python
def implot(img, img_filt, nomefiltro, kernelsize):
  plt.figure(figsize=(12,12))
  plt.subplot(1,2,1)
  plt.imshow(img)
  plt.title("Imagem Original")
  plt.axis("off")

  plt.subplot(1,2,2)
  plt.imshow(img_filt)
  plt.title("Filtro " + nomefiltro + ' ' + kernelsize)
  plt.axis("off")

  plt.show()
```

```python
img = data.moon()
```

```python
img_filt = cv2.blur(img, (7,7))
```

```python
implot(img, img_filt, 'media', '7x7')
```

```python
img_filt = cv2.blur(img, (5,5))
```

```python
implot(img, img_filt, 'media', '5x5')
```

```python
kernel = np.array([
    [0, -1,  0],
    [-1, 5, -1],
    [0, -1, 0],
])
```

```python
img_filt = cv2.filter2D(img, -1, kernel)
```

```python
implot(img, img_filt, "sharpen", "3x3")
```
