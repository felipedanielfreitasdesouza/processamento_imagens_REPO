```python
def calcHist(img):

  img_shape1, img_shape2 = img.shape[0], img.shape[1]

  hist_arr = np.zeros(256, dtype=int)

  for px in img:
    hist_arr[px] += 1

  plt.bar(range(256), hist_arr)
```
